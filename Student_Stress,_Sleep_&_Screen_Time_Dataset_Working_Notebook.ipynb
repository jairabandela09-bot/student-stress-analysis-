```python

# Loading the data

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student_stress_sleep_screen.csv")
```

```python

# Part A: Histograms

# Exercise A1: Create a histogram of sleep_hours.
# Requirements: 20 bins, figure size 8 x 5, Title, X-axis label, Y-axis label
# Excercise A2: Add a custom edge colour, transparency, gridlines, and a nicer title

plt.hist(df['sleep_hours'], bins=20, alpha = 1.0, edgecolor = 'white')

plt.figure(figsize = [8, 5])
plt.title('Distribution of Sleep Hours Among Students')
plt.xlabel('Sleep Hours')
plt.ylabel('Number of Students')

# What happens when the value of bins changes?
# A bin is a bounded interval with a defined lower and upper limit.
# Changing the number of bins means you change the number of intervals represented by the graph.
# Bins allow us to count up how many things fall into that category.
```

```python

# Part B: Scatterplots

# Create Screen Time vs Sleep.
# Requirements: custom marker size, alpha, title, axis labels
# Then, experiment with different markers, figure sizes, and point sizes.
# Last, try colouring by a different variable.

plt.figure(figsize = (8, 6))
marker_style = 'o'



plt.scatter(df['screen_time_hours'],
            df['sleep_hours'],
            s = 30,
            alpha = 1.0,
            edgecolors = 'white')

plt.title('Hours of Screen Time vs. Hours of Sleep')
plt.xlabel('Hours of Screen Time')
plt.ylabel('Hours of Sleep')
```
